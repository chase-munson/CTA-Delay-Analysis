The goal of this notebook is to create a script to segment the data for use in visualizations in tableau for distribution.

Current goals:

* Clean up duplicate delay observations
* Rename lines. Y to Yellow for example
* Normalize date/times

```python
# Import required libraries
import pandas as pd
import sqlite3
```

```python
# Connect to the SQLite database and read the train_data table into a DataFrame
with sqlite3.connect('../data/cta_data_05_2026.db') as conn:
    df = pd.read_sql('SELECT * FROM train_data', conn)
# Display the first 10 rows of the DataFrame
df.head(10)
```

```python
# Display the unique values in the 'line' column
df.line.unique()
```

Based on the values in the line column, Brn, G, Org, P, and Y need to be renamed.

```python
# Clean up the 'line' column by replacing abbreviations with full names
df['line'] = df['line'].replace({
    'Brn': 'Brown',
    'G': 'Green',
    'Org': 'Orange',
    'P': 'Purple',
    'Y': 'Yellow',
})
```

```python
df.line.unique()
```

The next goal is to normalize the dates and time columns.

```python
df['timestamp'] = pd.to_datetime(
    df['timestamp'],
    format='ISO8601',
    utc=True
)
```
